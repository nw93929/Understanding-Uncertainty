For the following distributions:

You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.


### Logistic

The support is calculated by seeing what values are possible, aka when is the probability > 0. For logistic distribution, we see that the CDF never is 0, although it approaches it at the extreme $-\infty$ and goes to 1 for $\infty$, so therefore the support is the interval between these two values, $(-\infty,\infty)$.

The PDF is given by:
# ![](logderive.png)

```python
import scipy as sp

grid = np.linspace(-10, 10, 1000)

pdf = sp.stats.logistic.pdf(grid)
cdf = sp.stats.logistic.cdf(grid)

fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.lineplot(x=grid,y=pdf,ax=axes[0])
axes[0].set_title('Density')
sns.lineplot(x=grid,y=cdf,ax=axes[1])
axes[1].set_title('Distribution')
plt.show()
```

```python
sample = sp.stats.logistic.rvs(size=1000,random_state=10)
fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x=sample,ax=axes[0],label='kde')
sns.lineplot(x=grid,y=pdf,ax=axes[0],label='pdf',linestyle='--')
axes[0].set_title('Sample KDE')
sns.ecdfplot(x=sample,ax=axes[1],label='ecdf')
sns.lineplot(x=grid,y=cdf,ax=axes[1],label='cdf',linestyle='--')
axes[1].set_title('Sample ECDF')
plt.show()
```

The distributions of the sample and theoretical PDF and CDF is virtually the same.

```python
metabric = pd.read_csv('data/metabric.csv')
for col in metabric.columns:
    try:
        sns.kdeplot(metabric[col])
        plt.title(col)
        plt.show()
    except:
        pass
```

We see that Age at Diagnosis has a similar type of distribution from the Metabric cancer data.


### Exponential


The support is calculated by seeing what values are possible, aka when is the probability > 0. For exponential distribution, we see that the CDF never is 0 if x >= 0, since the second term is 1/something always and it is subtracted from 1, but it is 0 for x<0, therefore we have the support $[0,\infty)$.

The PDF is given by:
# ![](expderive.png)

```python
grid = np.linspace(-10, 10, 1000)

pdf = sp.stats.expon.pdf(grid)
cdf = sp.stats.expon.cdf(grid)

fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.lineplot(x=grid,y=pdf,ax=axes[0])
axes[0].set_title('Density')
sns.lineplot(x=grid,y=cdf,ax=axes[1])
axes[1].set_title('Distribution')
plt.show()
```

```python
sample = sp.stats.expon.rvs(size=1000,random_state=10)
fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x=sample,ax=axes[0],label='kde')
sns.lineplot(x=grid,y=pdf,ax=axes[0],label='pdf',linestyle='--')
axes[0].set_title('Sample KDE')
sns.ecdfplot(x=sample,ax=axes[1],label='ecdf')
sns.lineplot(x=grid,y=cdf,ax=axes[1],label='cdf',linestyle='--')
axes[1].set_title('Sample ECDF')
plt.show()
```

Again the distributions of the CDF and PDF are virtually the same, although the peak of the PDF is a little shorter and has slightly longer tails.

### Negative Binomial

The support for this function appears to be $[0,n]$ where n is the number of "trials", since the function is only defined for these values, any negative values do not have a probability because it doesn't make sense to thinking about it logically.

```python
grid = np.linspace(-10, 30, 1000)

pdf = sp.stats.nbinom.pmf(grid, n=10, p=0.5)
cdf = sp.stats.nbinom.cdf(grid, n=10, p=0.5)

fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.lineplot(x=grid,y=pdf,ax=axes[0])
axes[0].set_title('Density')
sns.lineplot(x=grid,y=cdf,ax=axes[1])
axes[1].set_title('Distribution')
plt.show()
```

```python
sample = sp.stats.nbinom.rvs(size=1000,random_state=10,n=10,p=0.5)
fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x=sample,ax=axes[0],label='kde')
sns.lineplot(x=grid,y=pdf,ax=axes[0],label='pdf',linestyle='--')
axes[0].set_title('Sample KDE')
sns.ecdfplot(x=sample,ax=axes[1],label='ecdf')
sns.lineplot(x=grid,y=cdf,ax=axes[1],label='cdf',linestyle='--')
axes[1].set_title('Sample ECDF')
plt.show()
```

The CDF is virtually the same for the sample and theoretical, however the PDF differs greatly. In the theoretical it has a steep rise on the right side while it is more bellshaped in the sample.
